```python
import string
from nltk.tokenize import word_tokenize

text = "Jyoti is$$$$$ a@ good frnd of mine"

Aryanshi = word_tokenize(text)
print(Aryanshi)

words = []
for i in Aryanshi:
    if i not in string.punctuation:
        words.append(i)

print(words)
```

```python
import nltk
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize

nltk.download('stopwords')
nltk.download('punkt')

text = "Code is like humor: when you have to explain it, it’s bad."

J = word_tokenize(text)

words = []

for i in J:
    if i.lower() not in stopwords.words('english'):
        words.append(i)

print(words)
```

```python
# Spelling checking 1
!pip install pyspellchecker
from spellchecker import SpellChecker
spell = SpellChecker()
text = "i huv a gud, bd, and ugly situation"
Aryanshi = text.split()
correct = []
for i in Aryanshi:
    correct.append(spell.correction(i))
print(correct)
```
